-** data frame creation**
 _DataFrame creation is the process of creating a table-like data structure with rows and columns to store, organize, and analyze data efficiently using libraries like pandas._
-    -> **data frame **_A DataFrame is a 2-dimensional table with rows and columns used to store, organize, and manipulate data efficiently in Python using the pandas library._
-    -> **data write **
Data writing is the process of saving data to a file, database, or other storage so it can be accessed or used later._
- **delta format**
_Delta Format (Delta Lake) is a storage format built on top of Parquet that provides reliable and efficient data storage for big data processing. with features like ACID transactions, versioning, and support for updates and deletes._
-    ->** mode**
        _there are 6 types of modes which are 
        Delta → Advanced format with ACID transactions, history, and updates (mostly used in Databricks).
Parquet → Column-based format, fast for analytics and big data.
CSV → Simple row-column format, used for data exchange.
JSON → Used for semi-structured data and APIs.
Avro → Used for data streaming and schema-based storage.
ORC → Optimized format for Hive and big data queries.
and for write we have 4 modes , Append, Overwrite, Error, Ignore
_



```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.getOrCreate()

spark_df = spark.createDataFrame(df)

spark_df.write.format("delta").mode("overwrite").saveAsTable("batch_2.data.customers")
```
